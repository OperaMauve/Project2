# Mathematical Formulation
Suppose we have $n$ observations with $p$ features each.

Define
$$A\in M_{n, (p+1)}(\mathbb{R})$$
as the design matrix, where the first column is all ones to include and intercept.

The target vector is $$y \in \mathbb{R}^n$$

We assume a linear model
$$y = AX+\epsilon$$
where:
* $X \in \mathbb{R}^{p+1}$ contains the unknown parameters
* $\epsilon$ represents noise or unexplained variation.

We want to minimise the empirical square-error loss

$$J(X) = \frac{1}{2n}||AX-y||^2_2$$

This is an example of the least squares problem.
The optimisation question is the same, however now we give it a statistical intepretation of the problem.

Given the above equation, We have that
$$\begin{aligned}
J(X) &= \frac{1}{2n}||AX-y||^2_2 \\
&= \frac{1}{2n}(AX-y)^T(AX-y)\\
&= \frac{1}{2n}(X^TA^T-y^T)(AX-y) \\
&= \frac{1}{2n}(X^TA^TAX-y^TAX-X^TA^Ty+y^Ty) \\
&= \frac{1}{2n}(X^TA^TAX-2X^TA^Ty+y^Ty)\\
\end{aligned}$$

The goal is to find whether our problem is convex, and therefore make that local and global minimums are equilvalent

We calculate the gradient $\nabla J(X)$

$$\begin{aligned}
\nabla J(X) &= \nabla \frac{1}{2n}(A^TX^TAX-2A^TX^Ty+y^Ty) \\
&= \frac{1}{2n} \nabla (X^TA^TAX-2A^TX^Ty+y^Ty) \\
&= \frac{1}{2n} (\nabla (X^TA^TAX) -2 \nabla(y^TAX) + \nabla (y^Ty)) \\
&= \frac{1}{2n} (2A^TAX-2A^Ty) \\
&= \frac{1}{n} A^T(AX-y) \\
\end{aligned}$$

In the prior calculation, we may find the formula I've used by showing that they are equilvalent by differentiating term by term.

At a stationary point,
$$\nabla J(X) = 0$$
Thus,
$$ A^T(AX-y) = 0$$
yields,
$$ A^TAX = A^Ty$$

giving us the normal equations.



## Some formula used in matrix differentiation
Assuming that $A$ is symmetric
In our case $A' = A^TA$ is obviously symmetric, $A'^T = (A^TA)^T = A^TA = A'$
$$\nabla X^TAX = 2AX$$
$$\begin{aligned}
\nabla X^TAX &=
    \begin{bmatrix}
        \frac{\partial }{\partial x_1} \\
        \frac{\partial }{\partial x_2} \\
        \vdots \\
        \frac{\partial }{\partial x_{p+1}} \\
    \end{bmatrix}
    (
    \begin{bmatrix}
        x_1 & x_2 & \dots & x_{p+1} \\
    \end{bmatrix}
    \begin{bmatrix}
        A_{1,1} & A_{1,2} & \dots & A_{1,p+1} \\
        A_{2,1} & A_{2,2} & \dots & A_{2,p+1} \\
        \vdots & \vdots & \ddots & \vdots \\
        A_{p+1,1} & A_{p+1,2} & \dots & A_{p+1,p+1} \\
    \end{bmatrix}
    \begin{bmatrix}
        x_1 \\
        x_2 \\
        \vdots\\
        x_{p+1} \\
    \end{bmatrix}
    )
    \\
    &\text{ Evidently, this is going to be a scalar product of the $\nabla$ operator}
    \\
    &\text{ Lets denote $A_i$ as the i'th column of $A$}
    \\
    &= \nabla (X \cdot \sum_{i = 1}^{p+1} x_i A_i)
    \\
    &= \nabla (\sum_{j = 1}^{p+1} \sum_{i= 1}^{p+1} x_j x_i A_{i,j})
    \\
    & \text{Lets look at $\nabla X^TAX$ component wise}
    \\
\nabla X^TAX_{i} &=  \frac{\partial}{\partial x_i} (\sum_{j = 1}^{p+1} \sum_{k= 1}^{p+1} x_j x_k A_{k,j})
    \\
    &\text{Terms are non-zero when $i=j$ or $i = k$, Thus}
    \\
    &=\frac{\partial}{\partial x_i} (\sum^{p+1}_{k = 1}x_i x_k A_{k,i}+ \sum^{p+1}_{j = 1}x_j x_i A_{i,j}- x_i^2A_{i,i})
    \\
    &\text{Since $A$ is symmetric}
    \\
    &= 2\sum^{p+1}_{j = 1} x_k A_{k,i}
    \\
    &= 2AX_i
    \\
\end{aligned}$$

This is shown with much better clarity using einstein notation, Kronecker delta helps use handle the diagonal terms here.

Let $f(X) = X^TAX$, Then
$$\begin{aligned}
\frac{\partial f}{\partial x_k} &= \frac{\partial}{\partial x_k}(x_iA_{ij}x_j)
    \\
    &= \delta_{ij}A_{ij}x_j + x_iA_{ij}\delta_{jk}
    \\
    & = A_{kj}x_j + A_{ki}x_i
    \\
    &= 2A_{kj}x_j
    \\
    &= (2AX)_k
\end{aligned}$$

We also get that
$$\frac{\partial f}{\partial x_k} = A_{kj}x_j + A_{jk}x_j$$
meaning
$$\nabla_X(X^TAX) = (A+A^T)X$$

The other formula we used
$$ \nabla A^TX = A$$
is pretty self-explanatory, you can perform the calculation in your head

We calculate the Hessian $\nabla^2 J(X)$

From the previous section, we have that

$$ \begin{aligned}
\nabla^2 J(X) &= \nabla \nabla J(X) \\
& = \nabla \frac{1}{n})(A^T(AX-y)) \\
& = \frac{1}{n}(A^TA) \\
\end{aligned}$$

The Hessian does not depend on $X$ since $J$ is a quadratic function of $X$.

We want to show that $J$ is convex, so that we may say that the local minimum is the global minimum.
To show: $J$ is convex

From our Hessian, take any arbitrary $v \in \mathbb{R}^p$
Then,
$$ v^T\nabla^2J(X)v = \frac{1}{n} ||Av||^2_2$$

Geometrically speaking, this tells us that $J$ restricted on any path from $a$ to $b$ has non negative second derivative, which implies that $J$ is convex on the path.

Thus convex.
Because, the objective is convex, this stationary-point condition is not only necessary but sufficient for global property

When is the minimiser unique?

The minimiser is unique when $X$ has full column rank.

Just look at the Hessian.

# A Experiment with conditioning and stability


Generating random data

```python
rng = np.random.default_rng(42)

n = 100
x = rng.uniform(-1,1,n)
z = rng.normal(size = n)

X_true = np.array([1.0,2.0,-1.0])

def design_matrix(delta):
    return np.column_stack([
        np.ones(n),
        x,
        x + delta * z
    ])

def generate_y(A):
    noise = 1e-3 * rng.normal(size = n)
    return A @ X_true + noise
```

As we reduce the size of $delta$, the columns become identical, so $A$ is rank deficient

Solving using least squares

```python
deltas = [1, 1e-2, 1e-4, 1e-6]
for delta in deltas:
    A = design_matrix(delta)
    y = generate_y(A)

    X_est, *_ = np.linalg.lstsq(A, y, rcond=None)

    residual = np.linalg.norm(A @ X_est - y)
    error = np.linalg.norm(X_est - X_true)
    condition = np.linalg.cond(A)

    print(f"delta = {delta:.0e}")
    print(f"  condition number: {condition:.3e}")
    print(f"  residual norm:    {residual:.3e}")
    print(f"  parameter error:  {error:.3e}")
    print()
```

As demonstrated, when $\delta$ approaches 0, the condition number $\kappa(A)$ increases linearly.
The residual norm is small even when parameter error becomes large since the parameter error is the forward error and the residual norm is the backwards error. Intuitively, there is now a large range of acceptable parameters that can generate small norms. and when $\delta$ is zero, we have an extra dimension of parameters that can minimise backwards error.
